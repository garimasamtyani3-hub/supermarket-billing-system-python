```python
# Multiple item in a single transaction

items = []

while True:
    item_name = input("Enter item name: ")
    quantity = int(input("Enter quantity: "))
    price = float(input("Enter price per item: ₹"))

    total_price = quantity * price

    items.append({
        "name": item_name,
        "quantity": quantity,
        "price": price,
        "total": total_price
    })

    more = input("Do you want to add another item? (yes/no): ")

    if more.lower() == "no":
        break

print("\nItems Added:")
for item in items:
    print(item["name"], "-", item["quantity"], "x ₹", item["price"])

    
```

```python
# Total amount based on discount 

subtotal = 4500

if subtotal >= 5000:
    discount = subtotal * 0.15

elif subtotal >= 3000:
    discount = subtotal * 0.10

elif subtotal >= 1500:
    discount = subtotal * 0.05

else:
    discount = 0

print("Subtotal: ₹", subtotal)
print("Discount: ₹", discount)
```

```python
# Calculate subtotal , discount and final total

subtotal = 0

for item in items:
    subtotal += item["total"]

# Total based discount
if subtotal >= 5000:
    discount = subtotal * 0.15

elif subtotal >= 3000:
    discount = subtotal * 0.10

elif subtotal >= 1500:
    discount = subtotal * 0.05

else:
    discount = 0

final_total = subtotal - discount

print("\n----- BILL SUMMARY -----")
print("Subtotal       : ₹", round(subtotal, 2))
print("Discount       : ₹", round(discount, 2))
print("Final Total    : ₹", round(final_total, 2))
```

```python
# Itemized bill

print("\n" + "=" * 60)
print("              SUPERMARKET BILL")
print("=" * 60)

print(f"{'Item':<15}{'Qty':<8}{'Price':<12}{'Amount':<12}")

print("-" * 60)

for item in items:
    print(
        f"{item['name']:<15}"
        f"{item['quantity']:<8}"
        f"₹{item['price']:<11.2f}"
        f"₹{item['total']:<11.2f}"
    )

print("-" * 60)

print(f"{'Subtotal':<35} ₹{subtotal:.2f}")
print(f"{'Discount':<35} -₹{discount:.2f}")
print(f"{'Final Total':<35} ₹{final_total:.2f}")

print("=" * 60)
print("       THANK YOU FOR SHOPPING!")
print("=" * 60)
```
